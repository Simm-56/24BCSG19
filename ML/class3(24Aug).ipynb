## Adding tittles and labels

```python
import matplotlib.pyplot as plt
x = [5,8,10]
y = [12,16,6]

plt.title('Potting information')
plt.plot(x,y)
plt.ylabel('Y axis')
plt.xlabel('X axis')

plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.style
x = [5,8,10]
y = [12,16,6]

plt.title('Potting information')
plt.plot(x,y, color="red", linewidth=5)
plt.ylabel('Y axis')
plt.xlabel('X axis')
plt.savefig("Myplot.png", format="png")
plt.show()
```
